```python
tabela = pd.read_excel(r"C:\Users\douglas_paixao\Downloads\Vendas_WEG_Pecas_Industriais.xlsx")
display(tabela)
```

```python
print(tabela["Valor Final"].apply(type).value_counts())
```

```python
tabela ["Valor Final"] = (
    tabela ["Valor Final"]
    .str.replace ("R$", "", regex = False)
    .str.replace (".", "", regex = False)
    .str.replace (",", "", regex = False)
    .str.strip()
)
```

```python
tabela ["Valor Final"] = tabela ["Valor Final"].astype(float)

faturamento = tabela ["Valor Final"].sum()

print (faturamento)
```

```python
print(f"Faturamento total: R${faturamento:,.2f}")
```
