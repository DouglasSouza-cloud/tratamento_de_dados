```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
dataset = pd.read_csv (r'C:\Users\douglas_paixao\Documents\dados_automacao_weg_bruto.csv')

print("--- Primeiras Linhas do Dataset Bruto ---")
display(dataset)
```

```python
# Corrigir os cabeçalhos do sistema para os padrões técnicos da WEG
dataset.columns = ['id','temperatura','pressao','status','genero','consumo_kwh']

# Validar a alteração
dataset.head()
```

```python
# Agrupar e explorar a coluna de Status
agrupar_status = dataset.groupby (['status']).size()
print("--- Agrupamento por Status")
display(agrupar_status)
```

```python
# Agrupar e explorar a coluna de Status
agrupar_genero = dataset.groupby (['genero']).size()
print("--- Agrupamento por Gênero")
display(agrupar_genero)
```

```python
agrupar_genero.plot.bar(color='purple', rot=0)
plt.title('Distribuição por Gênero do Operador')
plt.xlabel('Gênero')
plt.ylabel('Quantidade')

# ATIVA AS LINHAS: 'y' significa horizontais, '--' deixa pontilhado, alpha controla a transparência
plt.grid(axis='y', linestyle='--', alpha=0.7, color='gray')

plt.show()
```
